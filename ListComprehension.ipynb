Case Study – String Analysis using List Comprehension **Instructions **

Create a string variable called text

Store any sentence or word inside the variable

Use list comprehension to:

Convert all characters into uppercase

Extract only vowels from the string

Extract only consonants from the string

Display:

Original string

Uppercase characters

Vowels list

Consonants list

String without spaces

Print the output in a clean format using print() statements

```python
# 1. Create a string variable and store a sentence
text = "Python Programming is Fun"

# 2. Use List Comprehension for operations

# Convert all characters into uppercase
uppercase_chars = [char.upper() for char in text]

# Extract only vowels from the string
vowels = [char for char in text if char.lower() in ('a', 'e', 'i', 'o', 'u')]

# Extract only consonants from the string
consonants = [char for char in text if char.isalpha() and char.lower() not in ('a', 'e', 'i', 'o', 'u')]

# Get string without spaces
no_spaces_list = [char for char in text if char != ' ']
string_without_spaces = "".join(no_spaces_list) # converting list back to a clean string


# 3. Display the output in a clean format

print(f"Original String:          {text}")
print(f"Uppercase Characters:     {uppercase_chars}")
print(f"Vowels List:              {vowels}")
print(f"Consonants List:          {consonants}")
print(f"String Without Spaces:    {string_without_spaces}")
```
