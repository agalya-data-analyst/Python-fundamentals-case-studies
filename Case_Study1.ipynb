```python
# 1. Using input() function to get student details from the user

student_name = input("Enter student name:")

# 2. Using type casting: int() for student age and float() for course fee with input()function

student_age=int(input("Enter student age:"))
course_name =input("Enter course name:")
course_fee =float(input("Enter course fee:"))

# 3.GST calculation
gst = 0.18 * course_fee
total_fee = course_fee + gst

# 4.Display all student details using:print(),f-string

#student details

print("/n-----Student details-----")

print(f"Student age :{student_name}")
print(f"Student age:{student_age}")
print(f"Course name:{course_name}")
print(f"Course fee:{course_fee}")

#Fees details

print(f"GST:{gst}")
print(f"Total fee:{total_fee}")

#Data types

print("/n-----Data types-----")

print(type(student_name))
print(type(student_age))
print(type(course_fee))
```
